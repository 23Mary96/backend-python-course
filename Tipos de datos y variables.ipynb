# Variables
#### ¿Qué es? 
-Son un espacio temporal para guardar un valor con un nombre específico
-No hay necesidad de declarar el típo de dato explísitamente
-Python es un lenguaje tipado dinamicamente, lo que quiere decir que el tipo de dato se define mientras corre el programa
-Las variables en python funcionan diferente que enotros lenguajes de programación
-No necesitamos crear las variables, solo asignamos un nombre a un valor usando el operados "designación" (=)
#### Tipos de Variables
-Variables globales: Son aquellas que se utilizarán para todo el programa
-Varibles locales: Son aquellas que se vana a utilizar solo en un bloque específico

## Variables Globales
-Declaramos variables globales afuera de funciones
-El valor de la variable puede ser constante para todo el programa
-Cualquier función podrá acceder a estas variables
-Podemos decir que las variables globales tienen un alcance que es igualmente global
-Normalmente son más usadas para constantes
-Las variables globales mantienen sus valores a través de la ejecusión completa de tu programa
-Una variable global puede ser accesada por cualquier función

```python
#Variable global
my_global_variable=50
def my_funcion_1():
    pass
def my_funcion_2():
    pass
```

## Variables Locales
-Usarlas dentro de una función
-Podemos tener una restricción al acceder a una variable local desde otra función
-No podemos usarlas durante todo el programa (estan limitadas a la función)
-El alacance y visibilidasd se limita al interior de la función
-Lo que dure la función será lo que dure el alcance de nuestra variable

```python
def my_funcion_1():
    my_variable=10 #Local
    print("my_funcion_1 -el valor de my_local_variable es:", my_variable)
    
def my_funcion_2():
    my_variable=20 #Local
    print("my_funcion_2 -el valor de my_local_variable es:", my_variable)

def my_funcion_3():
    print("my_funcion_3 -el valor de my_local_variable es:", my_variable)
```

## Palabra reservada "global"

```python
def my_funcion_1():
    my_variable=10 #Local
    print("my_funcion_1 -el valor de my_local_variable es:", my_variable)
    
def my_funcion_2():
    global my_variable
    my_variable=20 # Cambia a ser global por el hecho de poner "global"
    print("my_funcion_2 -el valor de my_local_variable es:", my_variable)

def my_funcion_3():
    print("my_funcion_3 -el valor de my_local_variable es:", my_variable)
```

### Forma binaria
Es un código organizado en bites, es un lenguaje, también conocido como el lenguaje de las máquina y tiene 1 y 0

#### Valor base 2
- Solamente acepta 2 dígitos, el 1 y el 0
- Usamos el prefijo "0b" o "OB"

```python
my_binari_number=0b11 #1*2^1+1*2^0
print("el valor de my_binary_number es:", my_binari_number)
```

```python
my_binari_number=0B110 #1*2^2+1*2^1+0*2^0
print("el valor de my_binary_number es:", my_binari_number)
```

### Forma Octal
#### El valor base es 8
- Solamente acepta dígitos del 0 al 7
- Usamos el prefijo "0o" o "0O"

```python
my_octal_number=0o17  #1*8^1 + 7*8^0
print("el valor de my_octal_number es:", my_octal_number)
```

```python
my_octal_number=0O153  #1*8^2 + 5*8^1 + 3*8^0
print("el valor de my_octal_number es:", my_octal_number)
```

## Forma Hexadecimal
#### El Valor base es 16
- Acepta dígitos del 0 al 9
- Acepta letras de la "Aa" a la "Ff"
- Usamos el prefijo "0x" o "0X"
- Representación, se comienza con los dígicos se comienza con las letras a=10, b=11 hasta f=15

```python
my_hexa_number=0x17  #1*16^1 + 7*16^0
print("el valor de my_hexa_number es:", my_hexa_number)
```

```python
my_hexa_number=0x1C  #1*16^1 + 12*16^0
print("el valor de my_hexa_number es:", my_hexa_number)
```
