 # Id ()

#### Función para obtener la dirección de memoria, devuelve un identificador único entero que representa la dirección de memoria donde se encuentra almacenado un objeto durante su vida en el programa

```python
id (p)
```

##### Objetos inmutables

```python
a=5
b=5
print (id(a))
print (id(b))
```

##### Objetos mutables

```python
x=[1,2,3]
y=x
print (id(x))
print (id(y))
```

```python
y .append (4)
print (x)
```

##### Copia de un objeto

```python
import copy
z=[1,2,3]
w= copy.copy (z)
print (id(z))
print (id(w))
```

#### Conversión de tipo

##### Convertir a entero (int)

```python
edad = int (input ("¿Cuantos años tienes?"))
print ("tendras", edad+1, "el proximo año")
```

##### Convertir a decimal (float)

```python
altura=float(input("¿Cuál es tu altura en metros?"))
print ("Tu altura es:", altura, "m")
```

#### Para extraer multiples valores

```python
a, b, c= input ("ingresa los 3 valores").split()
print("El valor de a es:", a)
print("El valor de b es:", b)
print("El valor de c es:",c)
```

```python
a, b, c= input ("ingresa los 3 valores separados por coma").split(",")
print("El valor de a es:", a)
print("El valor de b es:", b)
print("El valor de c es:",c)
```
