## Operadores de Asignación

¿Qué son los operadores de asignación?
- En python son usados para asognar valores a variables
- Normalmente incluyen al operador de asignación (=)

#### Tipos de Operadores de asignación
- "=": Operador de asignación
- "+=": Sumar y asignar
- "-=": Restar y asignar
- "*=": multiplicar y asignar
- "/=": dividir y asignar
- "//=": dividir con redondeo hacia abajo y asignar
- "%=": modulo y asignar
- "**=": exponente y asignar

```python
## "=": Operador de asignación
num1=20
num2=3
```

```python
## "+=": Sumar y asignar
num1=20
num2=3
print("num1 antes de la operación:", num1)
num1+=num2 #num1 se convertiá en num1 + num2
print("num1 después de la operación:", num1)
```

```python
## "-=": Restar y asignar
num1=20
num2=3
print("num1 antes de la operación:", num1)
num1-=num2 #num1 se convertiá en num1 - num2
print("num1 después de la operación:", num1)
```

```python
## "*=": multiplicar y asignar
num1=20
num2=3
print("num1 antes de la operación:", num1)
num1*=num2 #num1 se convertiá en num1 * num2
print("num1 después de la operación:", num1)
```

```python
## "/=": dividir y asignar
num1=20
num2=3
print("num1 antes de la operación:", num1)
num1/=num2 #num1 se convertiá en num1 / num2
print("num1 después de la operación:", num1)
```

```python
## "//=": dividir con redondeo hacia abajo y asignar
num1=20
num2=3
print("num1 antes de la operación:", num1)
num1//=num2 #num1 se convertiá en num1 // num2
print("num1 después de la operación:", num1)
```

```python
## "%=": modulo y asignar, va a ser el # que queda después de haber dividido (residuo)
num1=20
num2=3
print("num1 antes de la operación:", num1)
num1%=num2 #num1 se convertiá en el residuo de num1/num2
print("num1 después de la operación:", num1)
```

```python
## "**=": exponente y asignar
num1=20
num2=3
print("num1 antes de la operación:", num1)
num1**=num2 #num1 se convertiá en num1 a la num2
print("num1 después de la operación:", num1)
```

## Operadores Lógicos 

##### ¿Qué son los operadores lógicos? 
- Son usados principalmente para declaraciones condicionales
- Son indicadores estructurales
- Además de lógicos también se les conoce como conectores lógicos

##### Existen 3 tipos
- And (y) que regres True (verdadero) si todas las operaciones son verdaderas
- Or (o) que regresa True si por lo menos una de las operaciones es verdadera
- Not (no) regresa True si la operación es False (falso)

```python
### And (y) Todas las condiciones deben ser verdaderas
num1=11
num2=55
# num1 <= num2 =True
# num1!=num2 = True
print("si num1 <= num2 and num1!=num2", num1 <= num2 and num1!=num2)
```

```python
num1=11
num2=11
# num1 <= num2 =True
# num1!=num2 = False
print("si num1 <= num2 and num1!=num2", num1 <= num2 and num1!=num2)
```

```python
num1=11
num2=55
# num1 <= num2 =True
# num1!=num2 = True
# num1 >10 = True
print("si num1 <= num2 and num1!=num2", num1 <= num2 and num1!=num2 and num1>10)
```

```python
num1=11
num2=11
# num1 <= num2 =True
# num1!=num2 = False
# num1 >10 = True
print("si num1 <= num2 and num1!=num2", num1 <= num2 and num1!=num2 and num1>10)
```

```python
## Or (o) que regresa True si por lo menos una de las operaciones es verdadera
num1=11
num2=55
# num1 <= num2 =True
# num1!=num2 = True
print("si num1 <= num2 or num1!=num2", num1 <= num2 or num1!=num2)
```

```python
num1=11
num2=11
# num1 <= num2 =True
# num1!=num2 = False
print("si num1 <= num2 or num1!=num2", num1 <= num2 or num1!=num2)
```

```python
num1=11
num2=55
# num1 <= num2 =True
# num1!=num2 = True
# num1 >10 = True
print("si num1 <= num2 or num1!=num2  or num1>10", num1 <= num2 or num1!=num2 or num1>10)
```

```python
num1=11
num2=11
# num1 <= num2 =True
# num1!=num2 = False
# num1 >10 = True
print("si num1 <= num2 or num1!=num2 or num1>10", num1 <= num2 or num1!=num2 or num1>10)
```

```python
## Not (no) regresa True si la operación es False (falso)
num1=11
num2=55
# num1 <= num2 =True
print("si not num1 <= num2 ", not num1 <= num2)
```

```python
num1=11
num2=55
# num1!=num2 = True
print("si not num1 != num2 ", not num1 != num2)
```

## Operadores de identidad
¿Que són los operadores de identidad?
- Son usados para comparar la ubicacuión en memoria de 2 objetos
- Se utilizan para validar si los valores comparten el mismo espacio en memoria o no
- Regresa True/ Verdadero si lo comparten y False/Falso si no lo comparten
- Hay 2 operadores de identidad:
    - Is
    - Is not

```python
## is
num_1=11
num_2=55
print("num_1 is num_2", num_1 is num_2)
```

```python
num_1=77
num_2=77
print("num_1 is num_2", num_1 is num_2)
```

```python
## is not
num_1=11
num_2=55
print("num_1 is not num_2", num_1 is not num_2)
```

```python
num_1=77
num_2=77
print("num_1 is not num_2", num_1 is not num_2)
```

## Funciones básicas

### trunc()
-Se queda con la parte entera de los datos, no importa el decimal

```python
num1=64.99
num2=64.5
num3=64.51
num4=64.000001
print("num1:", num1)
print("num2:", num2)
print("num3:", num3)
print("num4:", num4)
```

```python
from math import trunc
num1=64.99
num2=64.5
num3=64.51
num4=64.000001
print("trunc(num1):", trunc(num1))
print("trunc(num2):", trunc(num2))
print("trunc(num3):", trunc(num3))
print("trunc(num4):", trunc(num4))
```

### round(), ceil() y floor()
- Diferentes tipos de redondeo

```python
## Round- redondeo normal
num1=64.99
num2=64.5
num3=64.51
num4=64.000001
print("round(num1):", round(num1))
print("round(num2):", round(num2))
print("round(num3):", round(num3))
print("round(num4):", round(num4))
```

```python
## Ceil= redondeo hacia arriba
from math import ceil
num1=64.99
num2=64.5
num3=64.51
num4=64.000001
print("ceil(num1):", ceil(num1))
print("ceil(num2):", ceil(num2))
print("ceil(num3):", ceil(num3))
print("ceil(num4):", ceil(num4))
```

```python
## floor -redondeo hacia abajo
from math import floor
num1=64.99
num2=64.5
num3=64.51
num4=64.000001
print("floor(num1):", floor(num1))
print("floor(num2):", floor(num2))
print("floor(num3):", floor(num3))
print("floor(num4):", floor(num4))
```

### Función %
- Para dar formato

```python
num1=64.99
num2=64.5
num3=64.51
num4=64.000001
print("'%.1f'%num1:", '%.1f'%num1)
print("'%.2f'%num2:", '%.2f'%num2)
print("'%.3f'%num3:", '%.3f'%num3)
print("'%.4f'%num4:", '%.4f'%num4)
```

```python
### format () -Más recomendado
num1=64.99
num2=64.5
num3=64.51
num4=64.000001
print("'{:.1f}'.format(num1):", '{:.1f}'.format(num1))
print("'{:.2f}'.format(num2):", '{:.2f}'.format(num2))
print("'{:.2f}'.format(num2):", '{:.2f}'.format(num2))
print("'{:.4f}'.format(num4):", '{:.4f}'.format(num4))
```
